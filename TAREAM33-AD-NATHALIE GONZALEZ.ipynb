## Análisis de crecimiento

Para analizar el crecimiento se utiliza el período 2013–2016,
correspondiente a los últimos cuatro años consecutivos con una
cobertura suficiente en el dataset. Los registros posteriores
son demasiado limitados para realizar una comparación consistente.

```python
# Filtrar únicamente los años 2013 y 2016
df_growth = df_year[df_year['Year'].isin([2013, 2016])].copy()

# Agrupar las ventas globales por género y año
genre_growth = df_growth.groupby(
    ['Genre', 'Year']
)['Global_Sales'].sum().unstack(fill_value=0)

# Verificar cómo quedó la tabla
genre_growth
```

```python
print(genre_growth.columns)
```

```python
genre_growth = genre_growth[
    (genre_growth[2013] > 0) &
    (genre_growth[2016] > 0)
].copy()

genre_growth['Growth_%'] = (
    (genre_growth[2016] - genre_growth[2013])
    / genre_growth[2013]
) * 100

genre_growth.sort_values('Growth_%', ascending=False)
```

```python
#mayor crecimiento
print("Género con mayor crecimiento:")

display(
    genre_growth.sort_values(
        'Growth_%',
        ascending=False
    ).head(1)
)
```

```python
#menor crecimiento
print("Género con menor crecimiento:")

display(
    genre_growth.sort_values(
        'Growth_%',
        ascending=True
    ).head(1)
)
```

```python
genre_growth['Growth_%'].sort_values().plot(
    kind='barh',
    figsize=(10,6)
)

plt.title('Crecimiento de ventas por género: 2013 vs 2016')
plt.xlabel('Crecimiento (%)')
plt.ylabel('Género')
plt.show()
```
